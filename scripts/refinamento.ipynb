```python
# Importando a biblioteca duckdb para criar o banco de dados em memória e a conexão com o banco de dados.
import duckdb
```

```python
# Vamos fazer o import do duckdb para criar o banco de dados em memória e a conexão 
# com o banco de dados utilizando o comando duckdb.connect() para criar a conexão com o banco de dados.
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
# Vamos selecionar os produtos para verificarmos como estão os nossos dados na camada gold até aqui.
df = con.execute('SELECT * FROM produtos').fetchdf()
df.head(10)
```

```python
# Agora eu vou criar uma Dimensão na camada Gold para os produtos, para isso eu vou criar uma tabela chamada dim_produtos e vou selecionar os campos id_produto, nome_produto, categoria_produto e preco_produto da tabela produtos e vou inserir esses dados na tabela dim_produtos.
# Mas antes vou delear as colunas id_categoria e id_fornecedor da tabela produtos, pois essas colunas não são necessárias para a criação da dimensão de produtos.
# O dataframe df2 vai ser igual ao df mas removendo as colunas id_categoria e id_fornecedor, para isso eu vou usar o método drop do pandas para remover essas colunas do dataframe df e criar um novo dataframe df2.
df2 = df.drop(columns=['id_categoria', 'id_fornecedor'])
df2.head(10)
# Pronto agora ele já tem uma estrutura sem dados duplicados e no modelo que queremos.
```

```python
# Agora sim vamos criar a nossa tabela dim_produtos na camada gold, para isso eu vou usar o comando CREATE TABLE IF NOT EXISTS para criar a tabela dim_produtos e depois vou usar o comando INSERT INTO para inserir os dados do dataframe df2 na tabela dim_produtos.

con.execute('''
CREATE TABLE IF NOT EXISTS dim_produtos (
    id_produto BIGINT,
    nm_produto TEXT,
    VL_PRODUTO FLOAT
            )
''')
```

```python
# Repare que a minha coluna VL_PRODUTO ficou com letra maiúscula, vou alterar para ficar com letra minúscula, para isso eu vou usar o comando ALTER TABLE para alterar o nome da coluna VL_PRODUTO para vl_produto.
con.execute('''ALTER TABLE dim_produtos RENAME COLUMN VL_PRODUTO TO vl_produto''')
```

```python
# Agora vou consultar essa tabela para a nossa dim_produtos para verificar se os dados foram inseridos corretamente.
df_dim = con.execute('SELECT * FROM dim_produtos').fetchdf()
df_dim.head(10)
```

```python
# Vamos agora fazer a inserção dos dados do dataframe df2 na tabela dim_produtos, 
# para isso eu vou usar o comando INSERT INTO para inserir os dados do dataframe df2 na tabela dim_produtos.
con.execute('''INSERT INTO dim_produtos SELECT * FROM df2''')
```

```python
# Agora eu preciso fchar a conexão com o banco de dados para liberar os recursos utilizados pela conexão,
# para isso eu vou usar o comando con.close() para fechar a conexão com o banco de dados.
# Se não efetuar o procedimento de fechamento do banco, e caso precisamos efetuar uma outra conexão 
# com o banco de dados, pode ocorrer um erro de conexão,
# pois o banco de dados pode estar bloqueado por conta da conexão anterior que não foi fechada corretamente.
con.close()
```
