O que são blocos de código.**

***BLOCOS DE CÓDIGO*** -
Em python se eu quiser criar uma coisa com mais de uma linha, eu vou ter que ***identar.***
Toda vez que eu preciso organizar as linhas em bloco de código, eu vou colocar o **: dois pontos**
vou dar um **[enter]**, automaticamente perceba que o cursor vai um **pouco para a direia.**
Eu posso também ajustar dando um [tab] do meu teclado.
O print imprime qualquer coisa!
--> Porém esse print aqui é **exclusivamente/pertence** do **if 0==0:**, pois depois do **: pontos**
ele tem que **respeitar uma hierarquia/bloco de código.**

A **identação em Python** é o **espaço em branco no início de uma linha de código**, e serve para definir blocos de código. É uma parte obrigatória da linguagem — diferente de outras linguagens como C, Java ou JavaScript, onde você usa {} para delimitar blocos, no Python isso é feito com a **indentação.**

Para que serve?

**A indentação define quais comandos pertencem ao mesmo bloco.**

Por exemplo, quando você cria uma **função, um if, for, while, etc., o que estiver identado será considerado parte daquele bloco.**

Exemplo:

if 10 > 5:

    print("10 é maior que 5")

    print("Esse bloco faz parte do IF")

print("Esse não faz parte do IF")

Nesse exemplo:

As duas primeiras print() estão dentro do bloco if, porque estão indentadas (espaçadas).

A última print() está fora do if, porque não está indentada.

Como fazer a indentação?

Normalmente se usa 4 espaços ou 1 tabulação (mas nunca misture os dois).

O PEP 8 (guia de estilo do Python) recomenda 4 espaços por nível de indentação.

O que acontece se não usar?

O Python vai gerar erro. Exemplo:

if 10 > 5:
print("Erro!")  # ERRO de indentação
Resultado:

IndentationError: expected an indented block

**Resumo:**

**Indentação** = espaços que indicam o nível do bloco de código.

É obrigatória no Python.

Serve para organizar e estruturar o código.

Ajuda a manter o código legível e bem organizado.

**O que são LOOPS ou Laços de Repetição?**

Você quer fazer uma coisa **mais de uma vez**.

Você deseja fazer uma **coisa várias vezes.**

→ ***Loop é eu repetir uma coisa UMA, DUAS, VÁRIAS VEZES o quanto for necessário.***

***Loop, ou laço, é uma estrutura de controle em programação que executa repetidamente um bloco de código.***
 A repetição ocorre até que uma condição especificada seja alcançada.
Os loops são estruturas fundamentais em programação e podem ser utilizados para automatizar tarefas.

→ Eles são úteis quando você quer executar uma **mesma tarefa várias vezes sem precisar copiar o mesmo código várias vezes.**

Existem três tipos principais de loops nas linguagens de programação mais comuns:

**for** – usado quando você **sabe exatamente** quantas vezes quer repetir algo.

Exemplo em Python:

for i in range(5):

    print("Olá", i)

**while** – usado quando você **não sabe** quantas vezes vai repetir, mas sabe a condição que precisa ser verdadeira para continuar.
Exemplo:

x = 0

while x < 5:

    print("Contando", x)

    x += 1

**do while** (não existe em Python, mas em outras linguagens como C, Java, etc.) – parecido com o while, mas garante que o bloco será executado pelo menos uma vez, mesmo que a condição seja falsa logo depois.

***Onde os loops são úteis?***

1. Para percorrer listas ou arquivos;

2. Para repetir cálculos;

3. Para criar menus interativos;

4. Para simular comportamentos repetitivos.

**Como utilizar While.**

Toda vez em um código qualquer que eu precisar repetir alguma coisa, seja um cálculo, seja uma palavra, uma expressão, uma função,
uma consulta ao banco de dados, toda vez que eu precisar ***repetir algo mais de uma vez***, eu tenho **DUAS FORMAS DE FAZER.**

1. Eu posso criar DUAS LINHAS de CÓDIGO ou QUANTAS LINHAS DE CÓDIGOS FOR NECESSÁRIO.
2. Ou eu posso criar simplesmente um LOOP.

Vamos fazer um laço de repetição, utilizando o While

Antes de mais nada eu crio um Contador com o valor de 1

Depois eu crio um laço de repetição while (Enquanto) o Contador que é a variável Contador for <=10, exibe o valor do contador e vai adicionando mais 1 durante cada interação até chegar no valor 10 

Aqui o contador começou com o valor 1

Enquanto o while contador for <=10 vai printando o valor

Lembrando que a estrutura do python é seguida por identação, ou seja, aqui o que está ‘dentro’ do while ele valida, identação é o espaço a frente na linha.

Cuidado! Você tem que colocar o incremento do contador, que é o contator +=1, por que só assim ele vai incrementando valor por valor.


```python
contador = 1

while contador <=10:
    print(contador)
    contador += 1
```

Vamos utilizar um outro Laço de Repetição que é o FOR

Aqui ele fez, para/FOR cada ‘i’ em/in na lista [1,2] adiciona cada item ao  ‘i’.

E para/FOR e cada vez estou pedindo para imprimir.

Aqui ele tem dois itens que é a lista [1,2] Le vai passar um item por vez e vai executar as ações do bloco de código aqui dentro, no caso imprimir. 

```python
for i in [1,2]:
    print(i)
```

Quando queremos fazer um teste ou utilizar um intervalo fictício, também podemos utilizar a função de range(1, 10)
Aqui com a função range(1, 10) ele vai criar de 1 a 10 registros e vamos imprimir isso em tela do 1 ao 9 ele imprime até antes do 10, se fosse do 1, 11 ele iria imprimir até 10:

```python
for i in range(1, 11):
    print(i)
```

```python
for i in range(1, 11):
    print(f'valor: {i}')
```

Outra coisa que podemos fazer é, dentro do laço de repetições, colocar algumas condições:

Se ‘i’ for igual a ‘5’ eu vou usar o break

Aqui ele vai do 1, 2, 3, 4 e 5, mas quando chegar no 5 ele vai parar.

Ele validou aqui dentro do bloco de código o i é igual a 5, sim! Então para! Podia ser outra coisa também! Depende o que você programou dentro do bloco de código.

```python
for i in range(1, 11):
    print(f'valor: {i}')
    if i == 5:
        break
```

**Diferença da utilização do WHILE e o FOR**

Perceba o seguinte que, aqui eu não estou utilizando o incremento do i ou seja, i +=1, pois não precisa desse contador, por que o FOR ele trabalha de forma diferente, ele vai pegar cada registro que a gene tem aqui ‘para FOR’ cada ‘i ou item’ ou ‘registro’ vai ter uma lista de 5 registros por exemplo, vai pegar 1, 2, 3, 4, 5, vai pegar um por vez e aplicar na lógica abaixo. 

Já o WHILE ele vai repetir todas essas instruções ‘while contador <=10:’ até que essa condição se torne verdadeira com o auxílio do contador para efetuar o incremento contador +=1


```python
for i in range(1, 11):
    print(f'valor: {i}')
    if i == 5:
        continue
```

```python
# 1 é igual a 5, NÃO. Então ele foi imprimindo todos os valores que são diferentes de 5, ou seja, se for igual a 5 continue, ele pula a impressão do valor 5. 

for i in range(1, 10):
    if i == 5:
        continue
    else:
        print(f'valor: {i}')
```

```python
for i in range(1, 10):
    if i % 2 == 0:
        print(f'Par: {i}')
    else:
        print(f'Impar: {i}')
```

```python
# criando as tabelas

# bronze_pessoas

cur.execute("""
            CREATE TABLE IF NOT EXISTS bronze_pessoas (
              id INTEGER PRIMARY KEY AUTOINCREMENT,
              nome TEXT,
              idade INTEGER,
              salario REAL                  
            )
            """)

# bronze_vendas

cur.execute(""" 
            CREATE TABLE IF NOT EXISTS bronze_vendas (
              id INTEGER PRIMARY KEY AUTOINCREMENT,
              vendedor TEXT,
              valor REAL,
              data TEXT
      )
      """)
# Comitar os comandos

con.commit()
```

```python
# Fazendo a Inserção nessas duas tabelas.
# Fazendo a Inserção para a tabela bronze_pessoas

# Comando cur.executemany() - Para fazer a inserção de vários registros.
# Essa função executemany() permite que passamos uma lista [] e depois passamos os valores das linhas que serão os dados do INSERT
# """INSERT INTO bronze_pessoas (nome, idade, salarios) VALUES (?,?,?) """, [ ('xxxx','xxxx','xxxx')]

cur.executemany("""
                INSERT INTO bronze_pessoas (nome, idade, salario) VALUES (?,?,?) """,
                [
                    ('João',25, 2000.00),
                    ('Maria',30, 2000.00),
                    ('José',35, 2000.00),
                    ('Ana',40, 2000.00),
                    ('Pedro',45, 2000.00),
                    ]
                )
```

```python
# Vamos criar as nossas tabelas já tendo o prefixo de Fatos e Dimensões
# Criando a tabela dim_pessoas
cur.execute("""
            CREATE TABLE IF NOT EXISTS dim_pessoas (
                id INTEGER PRIMARY KEY,
                nome TEXT,
                faixa_etaria TEXT               
            )            
            """)
```

```python
# Criando a tabela fat_salarios
# Como é a tabela FATO alem da chave primaria temos que fazer o id da tabela dim_pessoa e pessoa_id 
# E fazer a chave estrangeira FOREIGN KEY para (pessoa_id) vai fazer referencia para a dim_pessoa na chave chamada id da tabela dim_pessoas

cur.execute("""
            CREATE TABLE IF NOT EXISTS fat_salarios (
                id INTEGER PRIMARY KEY,
                pessoa_id INTEGER,
                salario REAL,
                FOREIGN KEY (pessoa_id) REFERENCES dim_pessoas(id)                             
            )            
            """)
```

```python
# Criando a tabela fat_vendas

cur.execute("""
            CREATE TABLE IF NOT EXISTS fat_vendas (
                id INTEGER PRIMARY KEY,
                pessoa_id INTEGER, 
                total_vendas REAL,
                FOREIGN KEY (pessoa_id) REFERENCES dim_pessoas(id)             
                )
""")
```

```python
# Agora precisamos trabalhar com esses dados para conseguirmos popular as nossas tabelas.
# dim_pessoas vai ser igual ao df_pessoas, mas eu quero do meu dataframe somente as colunas id, nome e faixa_etaria, ou seja, eu quero filtrar 
# somente essas colunas para criar a minha dimensão de pessoas.

dim_pessoas = df_pessoas[['id','nome','faixa_etaria']]
print(dim_pessoas)
```

```python
# Fazendo agora com a fat_salarios, a minha tabela de fatos de salários, ela vai ser igual ao df_pessoas, mas eu quero somente as colunas id e salario, ou seja, 
# eu quero filtrar somente essas colunas para criar a minha tabela de fatos de salários.
# Depois vou fazer o seguinte: Quero renomear a coluna de id para pessoa_id, para que eu consiga fazer a relação entre a minha tabela de fatos e a minha tabela de dimensão, ou seja,
# a minha tabela de fatos tem que ter a chave estrangeira que faz referencia a chave primária da minha tabela de dimensão. 

fat_salarios = df_pessoas[['id','salario']].rename(columns={'id':'pessoa_id'})
```

```python
# Vamos adicionar uma validação 
# Vamos criar a tablea fat_vendas que vai ser igual ao df_vendas.groupby para agrupar por vendedor e chamamos a função aggregate para agregar o valor total de vendas, ou seja,
# a soma do valor das vendas, ou seja, o total de vendas por vendedor. 
# agora fazemos o reset_index() para resetar o índice e transformar o vendedor em uma coluna normal, ou seja, ele não vai ser mais o índice do dataframe e sim uma coluna normal, 
# ou seja, ele vai ser um atributo da tabela de fatos.
# agora vamos renomear passando as colunas de vendedor : para nome e valor para : total_vendas
# agora vamos fazer a junção de merge dim_pessoas quando a gente tiver o nome, aqui ele vai fazer a junção pelo nome, ou sej,a um JOIN entre a tabela de fatos e a tabela de dimensão, 
# ou seja, ele vai pegar o nome do vendedor da tabela de fatos e vai comparar com o nome da tabela de dimensão, 
# quando ele encontrar ude 'id' com o 'total_vendas'
# e por fim vamos renomear a coluna 'id' para pessoa_id para que a gente consiga fazer a relação entre a tabela de fatos e a tabela de dimensão, ou seja,
# a chave estrangeira da tabela de fatos tem que ter o mesmo nome da chave primária da tabela de dimensão para que a gente consiga fazer essa relação.
fat_vendas = (df_vendas
              .groupby('vendedor')
              .agg({'valor': 'sum'})
              .reset_index()
              .rename(columns={'vendedor':'nome', 'valor':'total_vendas'})
              .merge(dim_pessoas, on='nome')[['id','total_vendas']]
              .rename(columns={'id':'pessoa_id'}))
```

Agora nos temos a nossa dimensão de pessoas dim_pessoas, temos a nossa fato salario fat_salario e a nossa fato e vendas fat_vendas.

```python
# Vamos salvar essas tabelas no banco de dados, ou seja, vamos fazer o processo de Insert utilizando o to_sql para cada uma dessas tabelas, 
# ou seja, a dim_pessoas, fat_salarios e fat_vendas
dim_pessoas.to_sql('dim_pessoas', con, if_exists='replace', index=False)
fat_salarios.to_sql('fat_salarios', con, if_exists='replace', index=False)
fat_vendas.to_sql('fat_vendas', con, if_exists='replace', index=False)
```
